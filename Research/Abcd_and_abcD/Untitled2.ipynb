```python
# 1. 準備: 多項式環と代数体の定義
x = polygen(QQ)
f = x^4 -3*x^3 +5*x^2 -3*x + 1
K.<a> = NumberField(f, 'a')

# 2. シンボリックな係数を定義
a0, a1, a2, a3 = var('a0, a1, a2, a3')

# 3. Kの元をシンボリックな多項式として表現
# この時点ではKの元ではない。
gamma_poly = a0 + a1*x + a2*x^2 + a3*x^3

# 4. Kの生成元aを1/aに移す自己同型写像を定義
gen_K = K.gen(0)
target_elem = 1 / gen_K
sigma = K.hom([target_elem],K)

# 5. 写像をgamma_polyに適用する
# K.hom(gen_K, target_elem)で得られる写像は、
# Kの元に対して作用するので、gamma_polyをKの元として扱う必要がある
# -> これは、xにgen_Kを代入してから写像を適用する、という流れと同じ

# 以下の2つのステップで実行する
# 5a. gamma_polyの変数xをaに置き換える
gamma_K = gamma_poly.subs({x: a})

# 5b. gamma_Kに写像sigmaを適用する
# この時点で、sigma(gamma_K) はKの元として計算される
sig_gamma_K = sigma(gamma_K)

# 6. 結果をaの多項式として整理する
# sig_gamma_K はすでにKの元なので、自動的にaの基底で表現される
# 係数を取り出すには .vector() を使う
sig_gamma_coeffs = sig_gamma_K.vector()

print(f"変換前の元 (シンボリック多項式): {gamma_poly}")
print(f"変換後の元 (Kの元): {sig_gamma_K}")
print(f"aの基底での係数: {sig_gamma_coeffs}")
```

```python
# 1. 代数体を定義する
x = polygen(QQ)
f = x^4 + 1
K.<alpha> = NumberField(f, 'alpha')

# Kの生成元を取得
gen_K = K.gen(0)

# 2. 自己同型写像 sigma を定義する
sigma = K.hom([1 / gen_K], K)

# 3. 不変な元の係数を求める
# 係数をシンボリック変数として定義
R.<c0, c1, c2, c3> = K['c0', 'c1', 'c2', 'c3']

# gammaをシンボリックな多項式として表現する
# Kの元ではないので、ここではシンボリックな表現を使う
gamma_poly = c0 + c1*gen_K + c2*gen_K^2 + c3*gen_K^3

# sigmaの作用を適用
# この時点ではKの元ではないので、直接は適用できない
# gamma_polyの各項にgen_Kとsigma(gen_K)を適用した結果を計算する
sig_gamma_poly = c0 + c1*sigma(gen_K) + c2*sigma(gen_K)^2 + c3*sigma(gen_K)^3

# sig_gamma_poly を alpha の多項式として展開する
# sigma(gen_K)はKの元なので、多項式に変換できる
sig_gamma_poly_expanded = sig_gamma_poly.polynomial(K.base_ring())

# gamma_poly を alpha の多項式として表現する
gamma_poly_K = K(gamma_poly) # ここでシンボリックな多項式をKの元に変換
gamma_poly_K_expanded = gamma_poly_K.polynomial(K.base_ring())

# 係数を比較して、連立方程式を立てる
coeffs_gamma = [gamma_poly_K_expanded.coefficient(alpha^i) for i in range(K.degree())]
coeffs_sig = [sig_gamma_poly_expanded.coefficient(alpha^i) for i in range(K.degree())]

# 係数が一致するという方程式
equations = [coeffs_gamma[i] == coeffs_sig[i] for i in range(K.degree())]

# 連立方程式を解く
solutions = solve(equations, [c0, c1, c2, c3])
print(solutions)

# 解の検証
# 解から具体的な係数を代入して、不変であることを確認する
if solutions:
    sol = solutions[0]
    c0_val = sol.rhs() if 'c0' in str(sol) else c0 # デフォルト値
    c1_val = sol.rhs() if 'c1' in str(sol) else c1
    c2_val = sol.rhs() if 'c2' in str(sol) else c2
    c3_val = sol.rhs() if 'c3' in str(sol) else c3

    # 不変な元の一般形
    invariant_form = c0_val + c1_val*alpha + c2_val*alpha^2 + c3_val*alpha^3
    
    # 検証
    is_invariant = sigma(invariant_form.subs(c1=-c3)) # c1 + c3 = 0 を仮定
    print(is_invariant)
    # これだけでは十分ではないので、解いた方程式の関係式で確認する
    
    # c1 = -c3 と c0, c2 を自由変数として、検証
    invariant_candidate = c0*alpha^0 + c1*alpha^1 + c2*alpha^2 - c1*alpha^3
    print(sigma(K(invariant_candidate)) == K(invariant_candidate))
```
