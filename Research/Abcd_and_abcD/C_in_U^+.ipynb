```python
def inv(s):
    return s[::-1].swapcase()
def conj(x,y):
    return y + x + inv(y)
    
#----------
def A(phi, n):
    if n >= 0:
        return 'C'*n + conj('a', phi)*n
    else:
        return 'c'*abs(n) + inv(conj('a', phi))*abs(n)
#----------
def St(phi, n):
    if n >= 0:
        return conj('a', phi)*n
    else:
        return inv(conj('a', phi))*abs(n)        
```

# 6_3, DbCa

```python
md = 'DbCa'

i = ideal(md)
print("-- Alex. ploly. --")
display(i.char_poly)
print("\n-- basis of units --")
ugens = i.UnitGroup.gens_values(); display(ugens)

print("\n-- C --")

R = 10
rng = range((-1)*R, R+1)
C = set()
for k,l in itertools.product(rng, rng):
    u = (ugens[0]^k)*(ugens[1]^l)
    C.add(u*i.tau(u))
display(len(C), list(C))
```

```python
print("\n-- generator --")
ig0 = i.generator
display(ig0)

phi = 'ecb' #'fecFb'

for n in range(-5,6):
    mc = A(phi, n) + md
    
    i = ideal(mc)
    ig = i.generator

    print(mc)
    display([n, i.matrix, i.eigen_vector, ig])
    print(f"Is {ig/ig0} in C?  --> {ig/ig0 in C}\n-----\n")
```
