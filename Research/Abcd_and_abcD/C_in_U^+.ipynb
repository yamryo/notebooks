### Preparation for functions

```python
def del_dup_and_sort(mlist:list):
    return sorted(list(set(mlist)), key=lambda x: x.polynomial().degree())

def compare_tuples(a,b):
    absed = [list(map(lambda x: abs(x), t)) for t in [a,b]]
    sums = [functools.reduce(lambda x,y: x^2+y^2, t) for t in absed]
    (A, B) = (a, b) if sums[0] == sums[1] else (sums[0], sums[1])
    if A > B: 
        return 1
    elif A < B: 
        return -1
    else: 
        return 0

def md_data(md:str, R:int=10):
    i = ideal(md)
    print("-- Alex. ploly. --")
    display(i.char_poly)
    print("\n-- basis of units --")
    ugens = del_dup_and_sort(i.UnitGroup.gens_values())
    display(ugens)
    
    print("\n==== C ====")
    print("\n-- basis of C --")
    Cgens = del_dup_and_sort([ug*(i.tau(ug)) for ug in ugens])
    display(Cgens)

    C = set()
    r = len(Cgens); print(r)
    signs = sorted(list(itertools.product(range(-1,2), repeat=r)), key=functools.cmp_to_key(compare_tuples))
    #---
    for ids in itertools.product(range(R+1), repeat=r):
        for s in signs:
            factors = [Cgens[i]**(s[i]*ids[i]) for i in range(r)]
            u = functools.reduce(lambda x,y: x*y, factors)
            if not (-1)*u in C:
                C.add(u)

    coeff_list = sorted([v.polynomial().coefficients() for v in C], key=functools.cmp_to_key(compare_tuples))
    for v in coeff_list:
        print(v)
```
