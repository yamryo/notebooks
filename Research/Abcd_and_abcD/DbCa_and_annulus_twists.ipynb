# 2. Annulus twist

```python
def inv(s):
    return s[::-1].swapcase()
def conj(x,y):
    return y + x + inv(y) if x else ''
#----------
def AnnulusTwist(phi, n):
    if n >= 0:
        return 'C'*n + conj('a'*n, phi)
    else:
        return 'c'*abs(n) + inv(conj('a'*abs(n), phi))
```

```python
phi = 'ecb' #'ecb' #'ECB' #'' #
#----------
def A63(n:int):
    return AnnulusTwist(phi, n) + monod63        
```

```python
for n in range(-1,5): #[-1,0]: #[1,-2]: #
    mc = A63(n)
    print(f"-- A^{n}(6_3) --")
    M = sp.block_collapse(rho2(mc))
    f = M.charpoly().as_expr().factor()
    display(mc, M, f)
    print("\n")
    #---
```

# 3. Conjugacy test function

```python
def transition_matrix(M,v):
    if type(v) == list:
        v = sp.Matrix(4,1,v)
    vecs = [(M**k)*v for k in range(5)]
    return sp.Matrix(4,4,lambda i,j: vecs[j][i])
tm = transition_matrix

def conjugation_matrix(M,N,v,w):
    P, Q = tm(M,v), tm(N,w)
    return P*(Q.inv())
```

```python
M, N = rho2(A63(0)), rho2(A63(3))

for v, w in itertools.combinations(itertools.product([0,1], repeat=4), 2):
    if v == (0,0,0,0) or w == (0,0,0,0):
         pass
    else: 
        v, w = list(v), list(w)
        S = tm(M,v)*(tm(N,w).inv())
        if is_Sp(S):
            display((v,w), S) #, N == S.inv()*M*S)
```
