```python
from sympy import Matrix

# 整数の行列を定義
M = Matrix([[1, 0, 2, 3],[3, 0, 1, 2],[2, 3, 1, 0],[0, 1, 2, 3]])
N = Matrix([[3, 0, 0, 1, 0],[0, 1, 0, -1, 2],[-3, 1, -1, 0, -3], [1, 2, -1, 0, 2], [-2, 0, 0, 2, 4]])
L = Matrix([[3, 0, -3, 1, -2],[0, 1, 0, -3, -2],[0, 0, -1, 0, 3], [1, 0, 0, -1, 2], [0, -1, 0, 0, 0]])

# 行列式の計算（戻り値も整数型）
for X in [M,N,L]:
    print(X.det())
```

```python
display(M,N,L)
L.minor(0,0), L.minor(3,0)
```

```python
L = Matrix([[3, 0, -4, 1, -2],[0, 1, 0, -3, -2],[0, 0, -1, 0, 3], [1, 0, 0, -1, 2], [0, -1, 0, 5, 0]])
display(L)

print(L.det())
L.minor(0,0), L.minor(3,0)
```

```python
from sympy import Symbol, symbols
x, y = symbols("x y")
x, y = 5, -1

L = Matrix([[x, 0, -4, 1, -2],[0, 1, 0, -3, 2],[0, 0, -1, 0, 3], [y, 0, 0, -1, 2], [0, -1, 0, 5, 0]])
display(L)

print(L.det())
L.minor(0,0), L.minor(3,0)
```
